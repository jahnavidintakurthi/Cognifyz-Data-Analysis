```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../dataset/restaurants.csv", engine="python")

print(df.head())
```

```python
# TASK 4 — RESTAURANT CHAINS

restaurant_counts = df["Restaurant Name"].value_counts()

chains = restaurant_counts[restaurant_counts > 1]

print("Restaurant Chains:")
print(chains.head(10))

chain_ratings = df.groupby("Restaurant Name")["Aggregate rating"].mean()

print("\nAverage Ratings of Chains:")
print(chain_ratings[chains.index].head(10))


chains.head(10).plot(kind="bar")

plt.title("Top Restaurant Chains")
plt.xlabel("Restaurant Name")
plt.ylabel("Number of Outlets")

plt.savefig("../images/restaurant_chains.png")

plt.show()
```
