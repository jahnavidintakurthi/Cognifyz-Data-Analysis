```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../dataset/restaurants.csv", engine="python")

print(df.head())
```

```python
# TASK 2 — CUISINE COMBINATION

cuisines = df["Cuisines"].dropna()


combination_counts = cuisines.value_counts().head(10)

print("Most Common Cuisine Combinations:")
print(combination_counts)


combo_ratings = df.groupby("Cuisines")["Aggregate rating"].mean()

print("\nCuisine Combination Ratings:")
print(combo_ratings.sort_values(ascending=False).head(10))


combination_counts.plot(kind="bar")

plt.title("Top Cuisine Combinations")
plt.xlabel("Cuisine Combination")
plt.ylabel("Count")

plt.savefig("../images/cuisine_combinations.png")

plt.show()
```
