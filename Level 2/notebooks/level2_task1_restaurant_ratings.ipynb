```python
import pandas as pd
import matplotlib.pyplot as plt

# LOAD DATASEt 
df = pd.read_csv("../../dataset/restaurants.csv", engine="python")

print(df.head())

# TASK 1 — RESTAURANT RATINGS

ratings = df["Aggregate rating"]

print("Rating Distribution:")
print(ratings.describe())

# Rating ranges
rating_ranges = pd.cut(ratings, bins=[0,1,2,3,4,5])

print("\nMost Common Rating Range:")
print(rating_ranges.value_counts())

# Average votes
average_votes = df["Votes"].mean()

print("\nAverage Votes Received:")
print(average_votes)

# Histogram
plt.hist(ratings)

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Ratings")
plt.ylabel("Number of Restaurants")

# SAVE IMAGE
plt.savefig("../images/ratings_distribution.png")

# SHOW GRAPH
plt.show()
```
