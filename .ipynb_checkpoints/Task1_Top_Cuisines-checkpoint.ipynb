```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("dataset/restaurants.csv", engine="python")

print(df.head())
```

```python
# TASK 1 — TOP CUISINES

cuisines = df["Cuisines"].dropna()

all_cuisines = []

for item in cuisines:
    split_items = item.split(",")

    for c in split_items:
        all_cuisines.append(c.strip())

cuisine_counts = pd.Series(all_cuisines).value_counts()

top_3 = cuisine_counts.head(3)

print("Top 3 Cuisines:")
print(top_3)

total_restaurants = len(df)

percentages = (top_3 / total_restaurants) * 100

print("\nPercentages:")
print(percentages)

top_3.plot(kind="bar")

plt.title("Top 3 Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Count")

plt.show()
```
