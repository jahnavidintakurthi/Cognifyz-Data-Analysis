```python
import pandas as pd
import matplotlib.pyplot as plt

# LOAD DATASET
df = pd.read_csv("../../dataset/restaurants.csv", engine="python")

# TASK 1 — TOP CUISINES

# Remove empty values
cuisines = df["Cuisines"].dropna()

# Store all cuisines
all_cuisines = []

# Split cuisines
for item in cuisines:
    split_items = item.split(",")

    for c in split_items:
        all_cuisines.append(c.strip())

# Count cuisines
cuisine_counts = pd.Series(all_cuisines).value_counts()

# Top 3 cuisines
top_3 = cuisine_counts.head(3)

print("Top 3 Cuisines:")
print(top_3)

# Percentage calculation
total_restaurants = len(df)

percentages = (top_3 / total_restaurants) * 100

print("\nPercentages:")
print(percentages)

# Bar chart
top_3.plot(kind="bar")

plt.title("Top 3 Cuisines")
plt.xlabel("Cuisine")
plt.ylabel("Count")

# Save image
plt.savefig("../images/top_cuisines.png")

# Show graph
plt.show()
```
