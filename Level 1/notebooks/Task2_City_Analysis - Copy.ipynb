```python
import pandas as pd
import matplotlib.pyplot as plt

# LOAD DATASET
df = pd.read_csv("../../dataset/restaurants.csv", engine="python")

# TASK 2 — CITY ANALYSIS

# Count restaurants by city
city_counts = df["City"].value_counts()

print("City with highest restaurants:")
print(city_counts.head(1))

# Average rating per city
avg_rating_city = df.groupby("City")["Aggregate rating"].mean()

print("\nAverage rating per city:")
print(avg_rating_city)

# City with highest average rating
highest_avg_rating = avg_rating_city.idxmax()

print("\nCity with highest average rating:")
print(highest_avg_rating)

# Plot top 10 cities
city_counts.head(10).plot(kind="bar")

plt.title("Top 10 Cities with Most Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")

# Save image
plt.savefig("../images/city_analysis.png")

# Show graph
plt.show()
```
